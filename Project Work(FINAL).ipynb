```python
import tkinter as tk
from tkinter import Label, Entry, Button, StringVar, OptionMenu, Text

class Currency_Converter:
    def __init__(self):
        self.root = tk.Tk()
        self.root.title('Currency converter')
        self.root.resizable(width=False, height=False)
        self.root.configure(background='white')
        
        self.rates = {
            'USD': 0.012,  
            'EUR': 0.011, 
            'GBP': 0.009,  
            'JPY': 1.79    
        }
        
        lable_one = Label(self.root, text="Indian Rupee", font="Times 15 bold", fg="GREEN", bg="WHITE")
        lable_one.grid(row=1, column=1)
        
        self.entry = Entry(self.root, bd=5, relief="groove", width=35)
        self.entry.grid(row=1, column=2)
        
        label_two = Label(self.root, text="Select Country", font="Times 15 bold", fg="RED", bg="WHITE")
        label_two.grid(row=2, column=1)
        
        self.selected_currency = StringVar(self.root)
        self.selected_currency.set('USD')
        currency_menu = OptionMenu(self.root, self.selected_currency, *self.rates.keys())
        currency_menu.grid(row=2, column=2)
        
        btn = Button(self.root, text="Convert", fg="black", font="Times 15 bold", bg="gray", command=self.currency_converter)
        btn.grid(row=3, column=1)
        
        btn = Button(self.root, text="Clear All", fg="black", font="Times 15 bold", bg="Gray", command=self.clear)
        btn.grid(row=4, column=1)
        
        self.result = Text(self.root, height=2, width=30, font="Times 10 bold", bd=5)
        self.result.grid(row=3, column=2)
        
        self.root.mainloop()
    
    def currency_converter(self):
        try:
            amount = float(self.entry.get())
            currency = self.selected_currency.get()
            rate = self.rates[currency]
            converted_amount = amount * rate
            
            self.result.delete(1.0, tk.END)
            self.result.insert(tk.END, f'{amount} INR = {converted_amount:.2f} {currency}')
        except ValueError:
            self.result.delete(1.0, tk.END)
            self.result.insert(tk.END, 'Please enter a valid number')
    
    def clear(self):
        self.entry.delete(0, tk.END)
        self.result.delete(1.0, tk.END)
        self.selected_currency.set('USD')

if __name__ == '__main__':
    Currency_Converter()
```
